```python
# Import all libraries which is related to your output
import pandas as pd # pandas we use for read the output
from sklearn.tree import DecisionTreeClassifier # scrikit-learn library used in machine learning
# and it draw decision tree and how to tree make decison as per your output
```

```python
# Create own dataset and variable name = data

data = {
    "Weather": [
        "Sunny", "Cloudy", "Sunny", "Cloudy", "Rainy",
        "Rainy", "Rainy", "Sunny", "Cloudy", "Rainy"
    ],
    "Temperature": [
        "Hot", "Hot", "Mild", "Mild", "Mild",
        "Cool", "Cool", "Hot", "Mild", "Mild"
    ],
    "Humidity": [
        "High", "High", "Normal", "High", "High",
        "Normal", "Normal", "High", "Normal", "High"
    ],
    "Wind": [
        "Weak", "Weak", "Strong", "Strong", "Strong",
        "Strong", "Weak", "Weak", "Weak", "Strong"
    ],
    "Play": [
        "No", "Yes", "Yes", "Yes", "No",
        "No", "Yes", "No", "Yes", "No"
    ]
}
```

```python
# DataFrame represents data in rows and columns

df = pd.DataFrame(data)
df
```

```python
#convert text data yo numebric data becuause machine can't understand ur text data
#this all process name is label encoding
#replace means text to numebric
df_encoded = df.replace({
    "Sunny": 0,"Cloudy": 1,"Rainy": 2,
    "Hot": 0,"Mild": 1,"Cool": 2,
    "High": 0,"Normal": 1,
    "Weak": 0,"Strong": 1,
    "No": 0,"Yes": 1
})
```

```python
df_encoded #for print ypur data
```

```python

# Separate input and output

X = df_encoded.drop("Play", axis=1)
# Here we drop the Play attribute.
# axis=1 means remove column, not row.

y = df_encoded["Play"]
# Here we use the Play attribute as y (output).

X
```

```python
X
```

```python
y
```

```python

# Train Decision Tree model

model = DecisionTreeClassifier()

model.fit(X, y)

# X = Weather, Temperature, Humidity, Wind
# y = output labels
```

```python

print("\n Model Trained ")

# It shows your model is trained for prediction
# because you fit the model using X and y
```

```python

# Make a sample prediction

sample = [[1, 1, 0, 0]] #here u can create ur own sample data for prediction

prediction = model.predict(sample)
```

```python

if prediction[0] == 1:
    result = "Play"
else:
    result = "Don't Play"

print(result)
```
