```python
import pandas as pd

data = {
    "Car": [
        "Honda City", "Hyundai Creta", "Maruti Swift",
        "Tata Nexon", "Toyota Innova", "Honda Amaze",
        "Kia Seltos", "Maruti Baleno"],

    "Company": ["Honda", "Hyundai", "Maruti", "Tata",
        "Toyota", "Honda", "Kia", "Maruti" ],

    "Year": [ 2022, 2023, 2021, 2023, 2022, 2020, 2024, 2021],

    "Price": [1200000, 1500000, 700000, 1100000,
        2200000, 650000, 1600000, 800000],

    "Mileage": [18, 17, 23, 19, 14, 20, 16, 22],

    "Fuel": ["Petrol", "Diesel", "Petrol", "Diesel",
        "Diesel", "Petrol", "Petrol", "Petrol" ],

    "Rating": [4.5, 4.3, 4.2, 4.4, 4.6, 4.0, 4.5, 4.1]
}

df = pd.DataFrame(data)

print(df)
```
