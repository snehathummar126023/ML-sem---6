```python
# Import Libraries
import pandas as pd
```

```python
# Create your data
data = {
    'Age': ['Youth', 'Youth', 'Middle', 'Senior', 'Senior',
            'Middle', 'Senior', 'Youth', 'Youth', 'Senior'],
    'Income': ['High', 'High', 'High', 'Medium', 'Low',
               'Low', 'Low', 'Medium', 'Low', 'Medium'],
    'Student': ['No', 'No', 'No', 'No', 'Yes',
                'Yes', 'Yes', 'No', 'Yes', 'Yes'],
    'Creditrating': ['Fair', 'Excellent', 'Fair', 'Fair', 'Fair',
                     'Excellent', 'Excellent', 'Fair', 'Fair', 'Fair'],
    'Buyscomputer': ['No', 'No', 'Yes', 'Yes', 'Yes',
                     'No', 'Yes', 'No', 'Yes', 'Yes']
}

```

```python
# Print all data using pandas
df = pd.DataFrame(data)
df
```

```python
# now and convert categorical to numberical (label encoding)

from sklearn.preprocessing import LabelEncoder

label_encoders = {}

for column in df.columns:
    le = LabelEncoder()
    df[column] = le.fit_transform(df[column])
    label_encoders[column] = le
```

```python
# Split Features & Target

X = df.drop('Buyscomputer', axis=1)  # Features
y = df['Buyscomputer']               # Target
```

```python
X
```

```python
y
```

```python

# Apply KNN Algorithm

from sklearn.neighbors import KNeighborsClassifier

# Create model
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X, y)
```

```python

print('now model is trained')
```

```python

new_data = pd.DataFrame({
    'Age': ['Youth'],
    'Income': ['Medium'],
    'Student': ['Yes'],
    'Creditrating': ['Fair']
})
```

```python

# Loop of each column
for col in new_data.columns:
    new_data[col] = label_encoders[col].transform(new_data[col])
```

```python

# ----- Predict using trained KNN model -----
prediction = knn.predict(new_data)
prediction
```

```python

# Decode your prediction to original label

final_result = label_encoders['Buyscomputer'].inverse_transform(prediction)

print("""Predicted class (encoded):
Please check the `Buyscomputer` to verify if it is correct.""", prediction)

print("""Predicted class (actual):
Please check the `Buyscomputer` to verify if it is correct.""", final_result[0])
```
